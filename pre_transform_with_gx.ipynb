## Data Validation

### Create Data Context

```python
from great_expectations.data_context import FileDataContext

context = FileDataContext.create(project_root_dir='./')
```

### Connect to Datasource

```python
# Create Datasource
datasource_name = 'csv-data-P2M3'
#datasource = context.sources.add_pandas(datasource_name)
datasource = context.get_datasource(datasource_name) #For retry when already created datasource

# Create Data Asset
asset_name = 'P2M3'
path_to_data = 'P2M3_Nadia_Syailendra_data_raw.csv'
#asset = datasource.add_csv_asset(asset_name)
asset = datasource.get_asset(asset_name)  #For retry when already created asset name

# Build batch request
batch_request = asset.build_batch_request()
```

### Create Expectation Suite

```python
# Creat an expectation suite
expectation_suite_name = 'expectation-P2M3-dataset'
#context.add_or_update_expectation_suite(expectation_suite_name)
context.get_expectation_suite(expectation_suite_name)  #For retry when already created expectation suite

# Create a validator using above expectation suite
validator = context.get_validator(
    batch_request = batch_request,
    expectation_suite_name = expectation_suite_name
)

# Check the validator
validator.head()
```

### Expectations

```python
# Expectation 1 : Column 'Transaction_ID' must be unique

validator.expect_column_values_to_be_unique('Transaction_ID')
```

```python
# Expectation 2 : Column 'Age' must be between 18 and 100

validator.expect_column_values_to_be_between(
    column='Age', min_value=18, max_value=100
)
```

```python
# Expectation 3 : Column 'Account_Type' must contain one of the following 3 things :
# Savings
# Business
# Checking

validator.expect_column_values_to_be_in_set('Account_Type', ['Savings', 'Business', 'Checking'])
```

```python
# Expectation 4 : Column 'Transaction_Amount' must in form of float

validator.expect_column_values_to_be_in_type_list('Transaction_Amount', ['float'])
```

```python
# Expectation 5: Column 'Transaction_Currency' must be 3 digit long.

validator.expect_column_value_lengths_to_equal(
    column='Transaction_Currency',
    value=3
)
```

```python
# Expectation 6: Column 'Customer_Email' contains '@','.' and alphanumeric in between '@' and '.'

validator.expect_column_values_to_match_regex(
    column="Customer_Email",
    regex = r"^[^@\s]+@[^@\s]+\.+[^@\s]+$"
)
```

```python
# Expectation 7: Column 'Customer_Name' must be between 2 to 40 characters

validator.expect_column_value_lengths_to_be_between(
    column='Customer_Name',
    min_value= 2,
    max_value= 40
)
```

### Save to Expectation Suite

```python
#Menyimpan ke Expectation Suite
validator.save_expectation_suite(discard_failed_expectations=False)
```

### Create Checkpoint

```python
checkpoint_1 = context.add_or_update_checkpoint(
    name = 'checkpoint_1',
    validator = validator,
)
```

```python
# Run a checkpoint

checkpoint_result = checkpoint_1.run()
```
